## 1. My lane (or freestyle) and why
Content Refresh Prioritization (Lane 1: Decay & Refresh) . Search traffic steadily degrades when high-performing URLs become stale. Rather than waiting for rankings to drop completely, machine learning helps detect early multi-signal decay patterns so editorial teams can proactively optimize and refresh the highest-impact content.

```python
import os, sys, subprocess
import pandas as pd
import numpy as np

# Clone repository into Colab environment if not present
REPO_URL = "https://github.com/ARHAM008/flyrank-ml-internship"
REPO_DIR = "flyrank-ml-internship"

if not os.path.isdir(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

# Load the exact dataset that ships with the repo
csv_path = os.path.join(REPO_DIR, "data/raw/content_refresh_anonymized.csv")
if not os.path.exists(csv_path):
    # Fallback to direct raw GitHub link
    csv_path = "https://raw.githubusercontent.com/ARHAM008/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(csv_path)

print(f"Total pages loaded: {len(df)}")
print(f"Columns available: {df.shape[1]}")
print(f"Total pages in starter dataset: {len(df)}")
```

## 2. The question: decision, action, cost of a wrong call

 * **Unit of Analysis:** A single URL (`page_id` / `url`) evaluated over historical search performance windows.
* **The Research Question:** Which declining high-potential pages are suffering from content staleness and will yield the highest incremental organic traffic recovery after an editorial refresh?
* **The Decision:** Rank-order every published URL by expected traffic gain to decide which 20-50 pages the content team should refresh this sprint.
* **The Action:** Editorial and SEO specialists update, rewrite, and republish the top-ranked priority pages.
* **Cost of a False Positive (Predict Refresh, but page didn't need it):** Writer hours and content budgets are wasted on healthy or unrecoverable pages, creating opportunity cost where decaying pages get neglected.
* **Cost of a False Negative (Missed a decaying page):** High-potential URLs quietly slide down search result ranks, causing permanent loss of compounding organic traffic.
* **Why ML helps over hand-crafted rules:** Heuristic thresholds (like `age > 300 days`) fail because staleness interacts non-linearly with impression velocity, CTR shifts, and ranking positions across diverse page categories.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# 1. Aging Content Share (> 180 days)
pct_aging = (df["content_age_days"] > 180).mean() * 100

# 2. Opportunity Pool (High Impressions with Low/Sub-optimal CTR)
med_imp = df["impressions_90d"].median()
med_ctr = df["ctr"].median()
pct_high_imp_low_ctr = ((df["impressions_90d"] > med_imp) & (df["ctr"] < med_ctr)).mean() * 100

# 3. Overall target/candidate balance
target_share = df["will_decay"].mean() * 100 if "will_decay" in df.columns else df.iloc[:, -1].mean() * 100

print(f"1. Aging inventory: {pct_aging:.1f}% of pages are older than 180 days.")
print(f"2. High-opportunity subset: {pct_high_imp_low_ctr:.1f}% of pages have above-median impressions but below-median CTR.")
print(f"3. Decaying/refresh candidate target rate: {target_share:.1f}% of total inventory.")
```
