## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: Precision@K (specifically Precision@20 and Precision@50).Metric Rationale: An editorial team has fixed operational capacity (e.g., bandwidth to refresh 20 to 50 articles per sprint). Overall Accuracy or ROC-AUC is misleading because content teams do not act on the entire catalog; they act strictly on the top-$K$ recommendations. Maximizing Precision@$K$ guarantees that the limited writer budget is spent on pages that truly need intervention.Secondary Metric: Recall@K / PR-AUC to monitor overall candidate capture and handle class imbalance.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Non-linear interactions: Fixed heuristic rules (such as content_age > 180 days or impressions < 500) miss the multidimensional signals of decay. A page can be old but remain top-ranked with high CTR, while a newer page might rapidly lose ranking due to query shift.

Continuous re-ranking: ML scores dynamically weight changes across CTR velocity, position movement, and baseline traffic, providing a continuous prioritization queue rather than a rigid binary cutoff.

Avoiding wasted resources: In our baseline evaluations, heuristic rules drop in precision across broader sets, whereas learned models capture subtle feature boundaries that preserve Precision@50.
